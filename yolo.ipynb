```python
import cv2
import os

def extract_all_frames(video_path, output_folder):  
    # 1. Create output directory if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # 2. Open the video file (using the video_path parameter)
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Error: Could not open video {video_path}")
        return

    frame_count = 0
    while True:
        # Read the next frame
        success, frame = cap.read()
        
        # If the frame was not successfully read, the video has ended
        if not success:
            break
        
        # 3. Save the current frame as an image inside the output folder
        frame_name = os.path.join(output_folder, f"frame_{frame_count:05d}.jpg")
        cv2.imwrite(frame_name, frame)
        
        frame_count += 1

    # Release the video capture object
    cap.release()
    print(f"Successfully extracted {frame_count} frames to '{output_folder}'.")

# Example usage (Using 'r' before paths to prevent Windows backslash escape errors):
video_file = r"C:\Users\Ishfa\OneDrive\Desktop\Python Game Folder\YOLO\british_highway_traffic.mp4"
output_dir = r"C:\Users\Ishfa\OneDrive\Desktop\Python Game Folder\YOLO\extracted_frames"

extract_all_frames(video_file, output_dir)
```

```python
import cv2
import os

def extract_one_frame_per_second(video_path, output_folder):  
    # 1. Create output directory if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # 2. Open the video file
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Error: Could not open video {video_path}")
        return

    # 3. Get the video's frame rate (FPS)
    fps = round(cap.get(cv2.CAP_PROP_FPS))
    if fps == 0:
        fps = 30  # Fallback to a default if OpenCV fails to read the metadata
        
    print(f"Detected Video FPS: {fps}. Extracting 1 frame every {fps} frames.")

    frame_count = 0
    saved_count = 0
    
    while True:
        success, frame = cap.read()
        
        if not success:
            break
        
        # 4. Only save the frame if it aligns with the 1-second interval
        if frame_count % fps == 0:
            # We use saved_count in the filename so they are numbered 0, 1, 2, 3...
            frame_name = os.path.join(output_folder, f"second_{saved_count:05d}.jpg")
            cv2.imwrite(frame_name, frame)
            saved_count += 1
        
        frame_count += 1

    cap.release()
    print(f"Successfully finished! Extracted {saved_count} frames to '{output_folder}'.")

# Paths (Using 'r' for Windows paths)
video_file = r"C:\Users\Ishfa\OneDrive\Desktop\Python Game Folder\YOLO\british_highway_traffic.mp4"
output_dir = r"C:\Users\Ishfa\OneDrive\Desktop\Python Game Folder\YOLO\extracted_frames"

extract_one_frame_per_second(video_file, output_dir)
```

```python
import os
import cv2
import math

video_path = "road.mp4"
output_folder = "extracted_frames"

# 1. Create the output folder if it doesn't exist
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# 2. Open the original video
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the input video.")
    exit()

# 3. Get original video properties
orig_fps = cap.get(cv2.CAP_PROP_FPS)

# Calculate how many frames to skip to get 1 frame per second
skip_interval = math.ceil(orig_fps)

frame_index = 0
saved_count = 0

print(f"Original FPS: {orig_fps:.2f}")
print(f"Saving 1 frame per second to folder: '{output_folder}'...")

# 4. Loop through the video frames
while True:
    ret, frame = cap.read()
    if not ret:
        break  # End of video
    
    # Check if the current frame lands exactly on the 1-second interval
    if frame_index % skip_interval == 0:
        saved_count += 1
        # Create a padded filename like 'frame_0001.jpg', 'frame_0002.jpg'
        filename = f"frame_{saved_count:04d}.jpg"
        file_path = os.path.join(output_folder, filename)
        
        # Save the frame as a JPEG
        cv2.imwrite(file_path, frame)
        
    frame_index += 1

# 5. Clean up
cap.release()

print(f"Finished! Successfully saved {saved_count} JPEG frames.")

```

```python
import os
import cv2
from ultralytics import YOLO

# 1. Define folder paths
input_folder = "extracted_frames"
output_folder = "car_detected_frames"

# 2. Initialize the Large YOLO model
model = YOLO("yolov8l.pt")

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# List all images in the folder and sort them
image_extensions = (".jpg", ".jpeg", ".png", ".PNG")
frame_files = sorted([f for f in os.listdir(input_folder) if f.lower().endswith(image_extensions)])

print(f"Processing {len(frame_files)} frames...")

# COCO vehicle class IDs: 2=car, 3=motorcycle, 5=bus, 7=truck
VEHICLE_CLASSES = [2, 3, 5, 7]

# 3. Loop through each extracted frame image file
for filename in frame_files:
    file_path = os.path.join(input_folder, filename)
    
    # Read the image frame using OpenCV
    frame = cv2.imread(file_path)
    if frame is None:
        continue
    
    # CRITICAL FIX: 
    # 'classes=VEHICLE_CLASSES' forces YOLO to only look for cars, buses, and trucks.
    # We keep 'conf=0.15' low to ensure it catches the top-down shapes of all vehicles.
    results = model(frame, conf=0.15, classes=VEHICLE_CLASSES, verbose=False)
    
    # Check if any vehicles were detected in this frame
    # results[0].boxes contains the detections for this single image
    if len(results[0].boxes) > 0:
        output_path = os.path.join(output_folder, f"detected_{filename}")
        
        # Plot the bounding boxes onto the frame (will only show vehicles now!)
        annotated_frame = results[0].plot()
        
        # Save to output folder
        cv2.imwrite(output_path, annotated_frame)
        print(f"Vehicle detected and saved: {filename}")
    else:
        print(f"No vehicles detected in: {filename}")

print(f"Finished processing! Cleaned vehicle frames saved to '{output_folder}'.")
```
